# 01. Data Loading and Preparation

In this notebook, we will load our raw data. We want to combine the training and testing sets into one big dataset. This makes data cleaning much easier later. We will also separate the house prices (`SalePrice`) into a different variable.

```python
import pandas as pd
```

## 1. Load Raw Data
Let's load the data from the `data/raw/` folder. We use the `Id` column as the index because it's just a row number, not a real house feature.

```python
X_train = pd.read_csv('../data/raw_data/train.csv', index_col="Id")
X_test = pd.read_csv('../data/raw_data/test.csv', index_col="Id")


print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
display(X_train.head(3))
```

## 2. Separate Target and Combine Data
The test data doesn't have house prices. So, we need to remove the `SalePrice` column from the training data and save it. After that, we join `X_train` and `X_test` together. This way, we can clean all the data at once in the next steps.

```python
y_train = X_train.pop("SalePrice").to_frame()

X = pd.concat([X_train, X_test])

print(f"Combined features X shape: {X.shape}")
print(f"Target variable y_train shape: {y_train.shape}")
```

## 3. Save Processed Data
Finally, we save our combined data into the `data/processed/` folder. We will use these new files in our next notebook.

```python
X.to_csv('../data/processed/all_X.csv')
y_train.to_csv('../data/processed/y_train.csv')
```
